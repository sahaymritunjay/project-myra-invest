```sql
CREATE OR REPLACE TABLE workspace.myra_invest.gold_stock_signals
USING DELTA
AS

SELECT

    *,

    CASE
        WHEN priceChangePct > 2 THEN "BUY"
        WHEN priceChangePct > 0 THEN "WATCH"
        ELSE "HOLD"
    END AS technicalSignal

FROM workspace.myra_invest.silver_stock_prices;

SELECT *

FROM workspace.myra_invest.gold_stock_signals

LIMIT 20;


SELECT

technicalSignal,
COUNT(*) totalStocks

FROM workspace.myra_invest.gold_stock_signals

GROUP BY technicalSignal;
```
