```sql

CREATE OR REPLACE TABLE workspace.myra_invest.silver_stock_prices
USING DELTA
AS
SELECT
    symbol,
    companyName,
    tradeDate,

    CAST(open AS DOUBLE) AS open,
    CAST(high AS DOUBLE) AS high,
    CAST(low AS DOUBLE) AS low,
    CAST(close AS DOUBLE) AS close,

    CAST(volume AS BIGINT) AS volume,

    ROUND(close - open, 2) AS priceChange,

    ROUND(
        try_divide(close - open, open) * 100,
        2
    ) AS priceChangePct,

    currency,
    source,

    current_timestamp() AS ingestionTimestamp

FROM workspace.myra_invest.bronze_stock_prices;

SELECT
    symbol,
    tradeDate,
    open,
    close
FROM workspace.myra_invest.bronze_stock_prices
WHERE open = 0 OR open IS NULL;

SELECT *
FROM workspace.myra_invest.silver_stock_prices
LIMIT 20;

SELECT

    COUNT(*) total_records,

    COUNT_IF(priceChange IS NULL) missing_price_change,

    COUNT_IF(priceChangePct IS NULL) missing_pct,

    AVG(priceChangePct) avg_return,

    MAX(priceChangePct) best_return,

    MIN(priceChangePct) worst_return

FROM workspace.myra_invest.silver_stock_prices;

SELECT

symbol,
companyName,
priceChangePct

FROM workspace.myra_invest.silver_stock_prices

ORDER BY priceChangePct DESC

LIMIT 10;

SELECT

symbol,
companyName,
priceChangePct

FROM workspace.myra_invest.silver_stock_prices

ORDER BY priceChangePct ASC

LIMIT 10;
```
