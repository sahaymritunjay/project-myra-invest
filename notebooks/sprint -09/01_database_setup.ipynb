```python
%sql
CREATE SCHEMA IF NOT EXISTS workspace.myra_invest;

USE workspace.myra_invest;

CREATE TABLE IF NOT EXISTS workspace.myra_invest.bronze_stock_prices (

  symbol STRING,
  companyName STRING,
  tradeDate DATE,

  open DOUBLE,
  high DOUBLE,
  low DOUBLE,
  close DOUBLE,

  volume BIGINT,

  currency STRING,
  source STRING,

  ingestionTimestamp TIMESTAMP

)
USING DELTA
PARTITIONED BY (tradeDate);


CREATE TABLE IF NOT EXISTS workspace.myra_invest.silver_stock_prices (

  symbol STRING,
  companyName STRING,
  tradeDate DATE,

  open DOUBLE,
  high DOUBLE,
  low DOUBLE,
  close DOUBLE,

  volume BIGINT,

  priceChange DOUBLE,
  priceChangePct DOUBLE,

  ingestionTimestamp TIMESTAMP

)
USING DELTA
PARTITIONED BY (tradeDate);

CREATE TABLE IF NOT EXISTS workspace.myra_invest.gold_stock_signals (

  symbol STRING,
  tradeDate DATE,

  close DOUBLE,

  dailyReturnPct DOUBLE,

  movingAvg5 DOUBLE,
  movingAvg20 DOUBLE,

  volatility5 DOUBLE,

  technicalSignal STRING,

  ingestionTimestamp TIMESTAMP

)
USING DELTA
PARTITIONED BY (tradeDate);

USE SCHEMA workspace.myra_invest;

CREATE VOLUME IF NOT EXISTS myra_data;

SHOW VOLUMES IN workspace.myra_invest;

display(dbutils.fs.ls("/Volumes/workspace/myra_invest/myra_data"))
```
